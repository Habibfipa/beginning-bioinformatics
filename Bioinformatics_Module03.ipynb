####**TASK 18.1** Modify line 8 to create a fasta file of the hominid sequences

````python
# Add more hominid sequences
sequences["P.troglodytes"] = "ACGTCACTCTCCTA"
sequences.update({
    "G.gorilla": "ACGTCCTCCTA",
    "H.erectus": "ACGTCACTCTGCTA"
})

# Create a FASTA file
with open("hominid_sequences.fasta", "w") as fasta_file:
    for key, value in sequences.items():
        fasta_file.write(">" + key + "\n")
        fasta_file.write(value + "\n")

# Display the FASTA-formatted content
with open("hominid_sequences.fasta", "r") as fasta_file:
    print(fasta_file.read())
````

####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

````python
from google.colab import files

# Upload the Rosalind dataset
uploaded = files.upload()
input_file = next(iter(uploaded))

with open(input_file, "r") as f:
    text = f.read().strip()

# Count each word with a dictionary
word_counts = {}

for word in text.split():
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Print each word and its count
for word, count in word_counts.items():
    print(word, count)
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
from google.colab import files

# Upload the Rosalind dictionary dataset
uploaded = files.upload()
input_file = next(iter(uploaded))

with open(input_file, "r") as f:
    words = f.read().split()

# Keep one copy of each word, preserving first appearance order
unique_words = list(dict.fromkeys(words))

# Use count() to count each word
for word in unique_words:
    print(word, words.count(word))
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
from google.colab import files
from Bio.SeqIO import parse
from Bio.SeqUtils import gc_fraction

# Upload the Rosalind GC-content dataset
uploaded = files.upload()
input_file = next(iter(uploaded))

best_id = ""
best_gc = -1

# Find the sequence with the highest GC content
for seq_record in parse(input_file, "fasta"):
    gc_percent = gc_fraction(seq_record.seq) * 100

    if gc_percent > best_gc:
        best_gc = gc_percent
        best_id = seq_record.id

# Rosalind requires the ID and GC percentage on separate lines
print(best_id)
print(best_gc)
````
